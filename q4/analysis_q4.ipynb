```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the function f(x,y)
def f(x, y):
    return -1000*y + 1000*np.exp(-x) - np.exp(-x)

# give the exact analytical solution of the ODE with the initial condition y(0) = 1 and also for y(0) = 0.5
def exact_sol(x):
    return np.exp(-x) -0.5*np.exp(-1000*x)

def exact_sol2(x):
    return np.exp(-x)

# Plot the analytical solution
x = np.linspace(0, 2, 10000)
y1 = exact_sol(x)
plt.plot(x, y1, label='y(0) = 0.5')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Analytical solution of the ODE with IC y(0) = 0.5')
plt.legend()
plt.show()



```

```python
import numpy as np
import os

def read_file(file_path):
    # Use genfromtxt to handle mixed data types and skip the first row
    data = np.genfromtxt(file_path, delimiter=None, skip_header=1)  # Auto-detect delimiter (space or tab)
    return data[:, 0], data[:, 1]  # Separate into x and y columns

# Define directory
directory = "resultsq4"

# Define parameters
methods = ["output_Explicit_Euler", "output_Implicit_Euler"]
initial_conditions = [0.500000, 1.000000]
step_sizes = [0.100000, 0.002000, 0.000100]

# Initialize lists
results_explicit_IC1_x = [[] for _ in step_sizes]
results_explicit_IC1_y = [[] for _ in step_sizes]
results_explicit_IC2_x = [[] for _ in step_sizes]
results_explicit_IC2_y = [[] for _ in step_sizes]
results_implicit_IC1_x = [[] for _ in step_sizes]
results_implicit_IC1_y = [[] for _ in step_sizes]
results_implicit_IC2_x = [[] for _ in step_sizes]
results_implicit_IC2_y = [[] for _ in step_sizes]

# Read files
for method in methods:
    for i, ic in enumerate(initial_conditions):
        for j, step in enumerate(step_sizes):
            file_name = f"{method}_step_{step:.6f}_IC_{ic:.6f}.txt"
            file_path = os.path.join(directory, file_name)

            if os.path.exists(file_path):
                print(f"Reading: {file_path}")  # Debugging output
                x_data, y_data = read_file(file_path)

                if method == "output_Explicit_Euler":
                    if i == 0:
                        results_explicit_IC1_x[j] = x_data
                        results_explicit_IC1_y[j] = y_data
                    else:
                        results_explicit_IC2_x[j] = x_data
                        results_explicit_IC2_y[j] = y_data
                else:
                    if i == 0:
                        results_implicit_IC1_x[j] = x_data
                        results_implicit_IC1_y[j] = y_data
                    else:
                        results_implicit_IC2_x[j] = x_data
                        results_implicit_IC2_y[j] = y_data
            else:
                print(f"File not found: {file_path}")

# Print results for verification
for j, step in enumerate(step_sizes):
    print(f"Explicit Euler (IC=0.5, step={step}): x={len(results_explicit_IC1_x[j])}, y={len(results_explicit_IC1_y[j])}")
    print(f"Explicit Euler (IC=1.0, step={step}): x={len(results_explicit_IC2_x[j])}, y={len(results_explicit_IC2_y[j])}")
    print(f"Implicit Euler (IC=0.5, step={step}): x={len(results_implicit_IC1_x[j])}, y={len(results_implicit_IC1_y[j])}")
    print(f"Implicit Euler (IC=1.0, step={step}): x={len(results_implicit_IC2_x[j])}, y={len(results_implicit_IC2_y[j])}")
```

```python
# Plot results for IC=0.5 of Explicit Euler with one step size = 0.1 with the exact solution
plt.plot(results_explicit_IC1_x[0], results_explicit_IC1_y[0], label='Explicit Euler (IC=0.5, step=0.1)')
y1 = exact_sol(results_explicit_IC1_x[0])
plt.plot(results_explicit_IC1_x[0], y1, label='Analytical solution')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Explicit Euler with IC=0.5 and step=0.1')
plt.ylim(0, 1)
plt.legend()
plt.show()
```
